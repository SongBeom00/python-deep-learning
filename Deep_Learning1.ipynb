```python
def convert_one_hot_sklearn(class_label) :
    encoder = LabelEncoder()
    labels = encoder.fit_transform(class_label)
    
    labels = labels.reshape(-1,1)
    oh_encoder = OneHotEncoder()
    oh_labels = oh_encoder.fit_transform(labels)
    return oh_labels.toarray()
```

```python
marvel_labels = ['아이언맨','캡틴아메리카','헐크','블랙위도우','스파이더맨','엔트맨']
ohe = convert_one_hot_sklearn(marvel_labels)
print(ohe)
print(type(ohe))
print(ohe.shape)

classes = [3,2,5,1,4,5,6,4]
ohe = convert_one_hot_sklearn(classes)
print(ohe)
print(type(ohe))
print(ohe.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
import numpy as np

a = np.array([1,2,3,4,5,7,8,4,6,1])

def One_hot_en(data)  :
    label_en = LabelEncoder()
    label_en = label_en.fit_transform(data)
    
    label_en = label_en.reshape(-1,1)
    one_hot_label = OneHotEncoder()
    one = one_hot_label.fit_transform(label_en)
    return one.toarray()
ohe = One_hot_en(a)
print(ohe)
```

```python
def AND(a,b) :
    input = np.array([a,b])
    weights = np.array([0.4,0.4])
    bias = -0.6
    value =np.sum(input*weights) +bias

    if value <= 0 :
        return 0
    else :
        return 1

```

```python
print(AND(0,0))
print(AND(0,1))
print(AND(1,0))
print(AND(1,1))
```

```python
def MAE(y,y_pred) :
    return np.mean(np.abs((y-y_pred)))
```

```python
y = np.array([-3,-1,-2,1,3,-2,2,5,3,3,-2,-1,2])
yhat = np.array([-3,-1,-5,0,2,3,5,2,2,5,3,-2,2])

print(len(y))
print(len(yhat))
x = list(np.arange(len(y)))

plt.scatter(x,y,color='b')
plt.plot(x,yhat,color='r')
plt.legend()
plt.grid()
```

```python
print(MAE(y,yhat))
```

```python
def MSE(y,y_pred) :
    return np.mean(np.square((y-y_pred)))
```

```python
y = np.array([-3,-1,-2,1,3,-2,2,5,3,3,-2,-1,2])
yhat = np.array([-3,-1,-5,0,2,3,5,2,2,5,3,-2,2])

print(len(y))
print(len(yhat))
x = list(np.arange(len(y)))

plt.scatter(x,y,color='b')
plt.plot(x,yhat,color='r')
plt.legend()
plt.grid()
```

```python
print(MSE(y,yhat))
```
